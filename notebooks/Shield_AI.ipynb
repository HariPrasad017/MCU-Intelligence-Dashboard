# MODULE 2
# Director Intelligence

```python
if menu == "Director Intelligence":
    st.header("Director Intelligence Module")
    st.write(
        "Analyze MCU director performance using total worldwide box office, average IMDb rating, and ROI."
    )

    director_df = df.groupby("director").agg(
        worldwide_box_office="sum",
        imdb_rating="mean",
        roi="mean"
    ).reset_index()

    top_director = director_df.sort_values(
        "worldwide_box_office", ascending=False
    ).iloc[0]

    st.metric(
        "Top Director by Total Worldwide Box Office",
        top_director["director"]
    )
    st.metric(
        "Total Worldwide Box Office",
        f"${top_director['worldwide_box_office']:,.0f}"
    )

    st.subheader("Director Performance Overview")
    st.dataframe(
        director_df.sort_values("worldwide_box_office", ascending=False)
    )

    st.subheader("Top 10 Directors by Worldwide Box Office")
    st.bar_chart(
        director_df.sort_values("worldwide_box_office", ascending=False)
        .head(10)
        .set_index("director")["worldwide_box_office"]
    )
```

# MODULE 3
# Franchise Health Index

```python
def get_franchise(title):

    title = title.lower()

    if "avengers" in title:
        return "Avengers"

    elif "spider-man" in title:
        return "Spider-Man"

    elif "iron man" in title:
        return "Iron Man"

    elif "thor" in title:
        return "Thor"

    elif "captain america" in title:
        return "Captain America"

    else:
        return "Other"
```

Apply:


```python
df["franchise"] = df["title"].apply(
    get_franchise
)
```

Health:

```python
franchise_df = df.groupby(

    "franchise"

).agg({

    "success_score":"mean",

    "imdb_rating":"mean",

    "worldwide_box_office":"mean"

}).reset_index()
```

Display:


```python
st.dataframe(
    franchise_df
)
```
